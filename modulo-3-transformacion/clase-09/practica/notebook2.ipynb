conexion y muestra de info

```python
import sqlite3
import pandas as pd

conexion = sqlite3.connect("rrhh.db")
df = pd.read_sql("SELECT * FROM empleados", conexion)
conexion.close()

print(df)
print(df.dtypes)
```

Buscamos los campos con valores null o nan

```python
print(df.isnull().sum())
```

calcular el valor de nulos en porcentajes


```python
porcentaje = (df.isnull().sum() / len(df) * 100).round(1)
print(porcentaje)
```

limpiar las filas de los registros con datos  nulos

```python
df_sin_nulos = df.dropna()
print(f"Filas originales: {len(df)}")
print(f"Filas sin nulos:  {len(df_sin_nulos)}")
```
