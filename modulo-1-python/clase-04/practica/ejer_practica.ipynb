```python
def limpiar_precio(valor):
    try:
        limpio = valor.strip().replace("$", "")
        return limpio
    except ValueError:
        return None
def decimal_punto(valor):
    try:
        
        posicion = valor.find(",")
       
        if (len(valor) - posicion + 1) == 2 :
            valor.replace(".","")
            valor.replace("," , ".")
            return valor  
        
           

        posicion = valor.find(".")
        print(posicion)
        if (len(valor) - posicion + 1) == 2 :
            valor.replace(".","")
        else:
            valor.replace(".","")
            return valor  
       
                 
    except ValueError:
        return None
    

precios_crudos = ["$45.000", "$1,250.50", " $8500 ", "N/A", "$32.000"]


precios_limpios = [limpiar_precio(p) for p in precios_crudos]
print(precios_limpios)
precios_separados = [decimal_punto(p) for p in precios_limpios]
print(precios_separados)
# [45000.0, 1250.5, 8500.0, None, 32000.0]

#precios_validos = [p for p in precios_limpios if p is not None]
#print(precios_validos)
# [45000.0, 1250.5, 8500.0, 32000.0]
```
