PRUEBA DE SCRAPPING PARA OBTENER DATOS

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
import time

# =====================================================
# CONFIGURACIÓN
# =====================================================

URLS = [
    # Reemplazar por URLs permitidas de reseñas públicas
    "https://www.guiaoleo.com.ar/buenos-aires/restaurants"
    
]

HEADERS = {
    "User-Agent": (
        "Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
        "AppleWebKit/537.36 (KHTML, like Gecko) "
        "Chrome/126.0 Safari/537.36"
    )
}

# =====================================================
# SCRAPING
# =====================================================

def extraer_resenas(url):
    print(f"Procesando: {url}")

    response = requests.get(url, headers=HEADERS)

    if response.status_code != 200:
        print(f"Error {response.status_code}")
        return []

    soup = BeautifulSoup(response.text, "html.parser")

    resultados = []

    # ADAPTAR SELECTORES SEGÚN EL SITIO
    reviews = soup.select(".review")

    for review in reviews:
        try:
            autor = review.select_one(".author")
            rating = review.select_one(".rating")
            comentario = review.select_one(".comment")
            fecha = review.select_one(".date")

            resultados.append({
                "url": url,
                "autor": autor.get_text(strip=True) if autor else "",
                "rating": rating.get_text(strip=True) if rating else "",
                "fecha": fecha.get_text(strip=True) if fecha else "",
                "comentario": comentario.get_text(" ", strip=True) if comentario else ""
            })

        except Exception as e:
            print("Error:", e)

    return resultados


# =====================================================
# EXPORTAR
# =====================================================

def main():

    datos = []

    for url in URLS:
        datos.extend(extraer_resenas(url))
        time.sleep(2)

    df = pd.DataFrame(datos)

    print(df.head())

    df.to_csv(
        "resenas_consumidores.csv",
        index=False,
        encoding="utf-8-sig"
    )

    print(f"\nReseñas obtenidas: {len(df)}")
    print("Archivo generado: resenas_consumidores.csv")


if __name__ == "__main__":
    main()
```
